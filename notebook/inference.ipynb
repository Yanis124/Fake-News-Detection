```python
import joblib
import os
import pandas as pd
```

```python
EXP_NUMBER = "1"
```

```python
# Chemins
model_path = f"../models/TF-IDF/{EXP_NUMBER}/Random_Forest_tfidf_model.pkl"
tfidf_s1_path = f"../models/TF-IDF/{EXP_NUMBER}/tfidf_s1_vectorizer.pkl"
tfidf_s2_path = f"../models/TF-IDF/{EXP_NUMBER}/tfidf_s2_vectorizer.pkl"

# Charger le modèle
model = joblib.load(model_path)

# Charger les TF-IDF vectorizers
tfidf_s1 = joblib.load(tfidf_s1_path)
tfidf_s2 = joblib.load(tfidf_s2_path)
```

```python
sentence1 = "The cat is sleeping on the sofa."
sentence2 = "A cat naps on a couch."

# Si tu as déjà une fonction de preprocessing
def preprocess_text(text, nlp):
    doc = nlp(text)
    tokens = [
        token.lemma_.lower()
        for token in doc
        if token.is_alpha and not token.is_stop and len(token) > 2
    ]
    return " ".join(tokens)

import spacy
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])

# Preprocessing
s1_clean = preprocess_text(sentence1, nlp)
s2_clean = preprocess_text(sentence2, nlp)
```

```python
X1 = tfidf_s1.transform([s1_clean])
X2 = tfidf_s2.transform([s2_clean])

# Selon ton pipeline, il est possible que tu combines X1 et X2
import scipy.sparse as sp
X_test = sp.hstack([X1, X2])
```

```python
pred_class = model.predict(X_test)
pred_proba = model.predict_proba(X_test)  # si le modèle supporte predict_proba

print("Predicted class:", pred_class[0])
if pred_proba is not None:
    print("Class probabilities:", pred_proba[0])
```
